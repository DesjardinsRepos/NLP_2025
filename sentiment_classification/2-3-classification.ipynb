```python
# ✅ Preprocessing Function
# This cell defines our configurable preprocessing function to apply on the data later.

import pandas as pd
import re
import string
from nltk.corpus import stopwords, wordnet
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize
from spellchecker import SpellChecker
from num2words import num2words
import nltk
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')

df = pd.read_csv("TweetSentiment.csv", encoding="ISO-8859-1")[["text", "sentiment"]]

contractions = {
    "don't": "do not", "doesn't": "does not", "can't": "cannot", "i'm": "i am",
    "you're": "you are", "he's": "he is", "she's": "she is", "it's": "it is",
    "we're": "we are", "they're": "they are", "isn't": "is not", "aren't": "are not",
    "wasn't": "was not", "weren't": "were not", "won't": "will not", "wouldn't": "would not",
    "couldn't": "could not", "shouldn't": "should not", "i've": "i have", "you've": "you have",
    "we've": "we have", "they've": "they have", "i'll": "i will", "you'll": "you will",
    "he'll": "he will", "she'll": "she will", "we'll": "we will", "they'll": "they will",
    "there's": "there is", "that's": "that is", "what's": "what is", "who's": "who is"
}

emoticon_dict = {
    r"(:-\)|:\)|=\)|:\]|=])": "SMILE",
    r"(;-?\)|;-?\])": "WINK",
    r"(:D|=D|;D)": "LAUGH",
    r"(:\(|:-\(|=\[|:\[)": "SAD",
    r"(:\/|:-\/)": "SKEPTICAL",
    r"(<3)": "HEART",
    r"(:3)": "CUTE",
    r"(:P|:p|:-P|:-p|=P)": "PLAYFUL",
    r"(:=)": "CONFUSED",
}

def expand_contractions_fun(text):
    pattern = re.compile(r'\b(' + '|'.join(re.escape(key) for key in contractions.keys()) + r')\b')
    return pattern.sub(lambda x: contractions[x.group()], text)

def reduce_elongation_fun(word):
    return re.sub(r'(.)\1{2,}', r'\1\1', word)

spell = SpellChecker(distance=1)
def fast_spell_correct(text):
    corrected_tokens = []
    for token in text.split():
        if token.isalpha():
            corrected = spell.correction(token)
            corrected_tokens.append(corrected if corrected else token)
        else:
            corrected_tokens.append(token)
    return " ".join(corrected_tokens)

def get_wordnet_pos(tag):
    if tag.startswith('J'):
        return wordnet.ADJ
    elif tag.startswith('V'):
        return wordnet.VERB
    elif tag.startswith('N'):
        return wordnet.NOUN
    elif tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

def preprocessing(
    df,
    text_col="text",
    overwrite_column="text",
    lowercase=False,
    expand_contractions=False,
    remove_urls=False,
    emoticon_normalization=False,
    detect_censored=False,
    remove_mentions=False,
    remove_punctuation=False,
    preserve_ellipsis=False,
    remove_numbers=False,
    convert_numbers=False,
    remove_non_ascii=False,
    reduce_elongation=False,
    remove_stopwords=False,
    stemming=False,
    lemmatization=False,
    spelling_correction=False,
    strip_multispace=False,
):
    stop_words = set(stopwords.words("english")) if remove_stopwords else set()
    stemmer = PorterStemmer() if stemming else None
    lemmatizer = WordNetLemmatizer() if lemmatization else None

    def clean_text(text):
        if lowercase:
            text = text.lower()
        if expand_contractions:
            text = expand_contractions_fun(text)
        if remove_urls:
            text = re.sub(r"http\S+|www\S+|https\S+", "", text)

        if emoticon_normalization:
            for pattern, token in emoticon_dict.items():
                text = re.sub(pattern, token, text, flags=re.IGNORECASE)

        if detect_censored:
            text = re.sub(r"\*{2,}", "CENSORED", text)
        if remove_mentions:
            text = re.sub(r"@\w+", "", text)

        if remove_punctuation:
            if preserve_ellipsis:
                text = text.replace("...", "ELLIPSISTOKEN")
            text = text.translate(str.maketrans("", "", string.punctuation))
            if preserve_ellipsis:
                text = text.replace("ELLIPSISTOKEN", "...")

        if convert_numbers:
            try:
                text = re.sub(r"\d+", lambda m: num2words(int(m.group())), text)
            except Exception:
                pass
        elif remove_numbers:
            text = re.sub(r"\d+", "", text)

        if remove_non_ascii:
            text = text.encode("ascii", errors="ignore").decode()
        if spelling_correction:
            text = fast_spell_correct(text)

        tokens = word_tokenize(text)

        if stemming:
            tokens = [stemmer.stem(word) for word in tokens]
        if lemmatization:
            pos_tags = nltk.pos_tag(tokens)
            tokens = [lemmatizer.lemmatize(word, get_wordnet_pos(tag)) for word, tag in pos_tags]
        if reduce_elongation:
            tokens = [reduce_elongation_fun(word) for word in tokens]
        if remove_stopwords:
            tokens = [word for word in tokens if word not in stop_words]

        text = " ".join(tokens)

        if strip_multispace:
            text = re.sub(r"\s{2,}", " ", text).strip()

        return text

    df[overwrite_column] = df[text_col].astype(str).apply(clean_text)
    return df
```

```python
# ✅ Balancing Dataset Function

from sklearn.utils import resample

def get_balanced_dataset():
    df = pd.read_csv("TweetSentiment.csv", encoding="ISO-8859-1")[["text", "sentiment"]]
    df.dropna(inplace=True)

    max_size = df["sentiment"].value_counts().max()
    df_balanced = pd.concat([
        resample(class_df, replace=True, n_samples=max_size, random_state=42)
        for _, class_df in df.groupby("sentiment")
    ])

    return df_balanced
```

```python
# ✅ Naive Bayes: Different Vectorizers
# This cell compares BoW, TF-IDF, Binary Vectorizer, BoW with N-grams and HashingVectorizer for a Naive Bayes model.

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer, HashingVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, recall_score, f1_score
from sklearn.metrics import classification_report

df_balanced = get_balanced_dataset()

X_train, X_test, y_train, y_test = train_test_split(
    df_balanced["text"], df_balanced["sentiment"],
    test_size=0.2, stratify=df_balanced["sentiment"], random_state=42
)

vectorizers = {
    "CountVectorizer (BoW)": CountVectorizer(),
    "TF-IDF": TfidfVectorizer(),
    "Binary Vectorizer": CountVectorizer(binary=True),
    "BoW with N-grams": CountVectorizer(ngram_range=(1, 2)),
    "HashingVectorizer": HashingVectorizer(n_features=10000, alternate_sign=False)
}

accuracy_scores = []
for name, vectorizer in vectorizers.items():
    print(f"Evaluating {name}...")
    X_train_vec = vectorizer.fit_transform(X_train) if hasattr(vectorizer, "fit_transform") else vectorizer.transform(X_train)
    X_test_vec = vectorizer.transform(X_test)

    model = MultinomialNB()
    model.fit(X_train_vec, y_train)
    y_pred = model.predict(X_test_vec)

    print(classification_report(y_test, y_pred, digits=4))
    acc = accuracy_score(y_test, y_pred)
    rec = recall_score(y_test, y_pred, average="macro")
    f1 = f1_score(y_test, y_pred, average="macro")

    # from sklearn.metrics import ConfusionMatrixDisplay
    # ConfusionMatrixDisplay.from_predictions(y_test, y_pred)

    accuracy_scores.append((name, acc))

print("Accuracy Scores for Different Vectorizers:")
print("========================================")
print(", ".join(f"{name}: {acc:.4f}" for name, acc in accuracy_scores))
```

```python
# ✅ Naive Bayes: Different Preprocessing (n-gram BoW)
# Since n-grams BoW performs best, we will compare preprocessing options using it as baseline.
# applying all options reduce performance by about 2%

import pandas as pd
import torch
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
import random
import numpy as np
import warnings
import gc

warnings.filterwarnings("ignore")

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

df = get_balanced_dataset()

def train_and_eval(df):
    X_train, X_test, y_train, y_test = train_test_split(
        df["text"], df["sentiment"],
        test_size=0.2, stratify=df["sentiment"], random_state=42
    )

    vectorizer = CountVectorizer(ngram_range=(1, 2))
    X_train_vec = vectorizer.fit_transform(X_train)
    X_test_vec = vectorizer.transform(X_test)

    model = MultinomialNB()
    model.fit(X_train_vec, y_train)
    y_pred = model.predict(X_test_vec)
    acc = accuracy_score(y_test, y_pred)

    return acc

options = [
    "lowercase", "expand_contractions", "remove_urls", "emoticon_normalization",
    "detect_censored", "remove_mentions", "remove_punctuation", "preserve_ellipsis",
    "remove_numbers", "convert_numbers", "remove_non_ascii", "reduce_elongation",
    "remove_stopwords", "stemming", "lemmatization", "spelling_correction", "strip_multispace"
]

acc_base = train_and_eval(df)

print(f"{'Base accuracy (n-grams)':30s}: {acc_base:.4f}")
print()
print(f"{'Preprocessing Option':25s} | Peformance Change")
print("-" * 50)

for opt in options:
    kwargs = {k: False for k in options}
    kwargs[opt] = True
    df_pre = preprocessing(df.copy(), **kwargs)
    df_pre.dropna(inplace=True)

    acc = train_and_eval(df_pre)
    delta = acc - acc_base

    print(f"{opt:25s} | {delta:+.4f}")

    # Actually do garbage collection because my RAM gets full
    # Why is this not collected automatically after losing scope??
    del df_pre, acc, delta, kwargs
    gc.collect()
```

```python
# ✅ Naive Bayes: Error Analysis
# This cell claculates the main metrics and confusion matrix. The next one visualizes errors sorted by confidence.

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, recall_score, f1_score, confusion_matrix, ConfusionMatrixDisplay
import numpy as np
import matplotlib.pyplot as plt

df_balanced = get_balanced_dataset()

X_train, X_test, y_train, y_test = train_test_split(
    df_balanced["text"], df_balanced["sentiment"],
    test_size=0.2, stratify=df_balanced["sentiment"], random_state=42
)

vectorizer = CountVectorizer(ngram_range=(1, 2))
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

model = MultinomialNB()
model.fit(X_train_vec, y_train)

y_pred = model.predict(X_test_vec)
y_proba = model.predict_proba(X_test_vec)
confidences = np.max(y_proba, axis=1)

acc = accuracy_score(y_test, y_pred)
rec = recall_score(y_test, y_pred, average="macro")
f1 = f1_score(y_test, y_pred, average="macro")
print(f"\nAccuracy: {acc:.4f}, Recall: {rec:.4f}, F1: {f1:.4f}")

errors_df = pd.DataFrame({
    "text": X_test.values,
    "true_label": y_test.values,
    "predicted_label": y_pred,
    "confidence": confidences
})
errors_df["correct"] = errors_df["true_label"] == errors_df["predicted_label"]
errors_only_df = errors_df[~errors_df["correct"]].sort_values(by="confidence", ascending=False)

cm = confusion_matrix(y_test, y_pred, labels=model.classes_)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=model.classes_)
disp.plot(cmap="Blues")
plt.title("Confusion Matrix: Naive Bayes with N-grams BoW")
plt.show()
```

```python
# ✅ Feed-Forward: Different Vectorizers (small NN)
# This cell compares different vectorizers for the smaller of two feed-forward neural networks.
# For 3 Epochs CountVectorizer performs best, but if we train the model to convergence (roughly 10 epochs), n-grams BoW perform best again

import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer, HashingVectorizer
from sklearn.metrics import accuracy_score, recall_score, f1_score, classification_report
import gc
import warnings
warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)

df_balanced = get_balanced_dataset()
df_balanced["label"] = df_balanced["sentiment"].astype("category").cat.codes

X_train, X_test, y_train, y_test = train_test_split(
    df_balanced["text"], df_balanced["label"],
    test_size=0.2, stratify=df_balanced["label"], random_state=SEED
)

class FeedForwardNN(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.fc = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.ReLU(),
            nn.Linear(128, 3)
        )
    def forward(self, x):
        return self.fc(x)

def train_and_eval_nn(X_train_tensor, y_train_tensor, X_test_tensor, y_test_tensor, input_dim):
    model = FeedForwardNN(input_dim)
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=0.001)
    model.train()
    for epoch in range(10):
        for i in range(0, len(X_train_tensor), 64):
            batch_x = X_train_tensor[i:i+64]
            batch_y = y_train_tensor[i:i+64]
            optimizer.zero_grad()
            outputs = model(batch_x)
            loss = criterion(outputs, batch_y)
            loss.backward()
            optimizer.step()
    model.eval()
    with torch.no_grad():
        preds = torch.argmax(model(X_test_tensor), dim=1)
        
    acc = accuracy_score(y_test_tensor, preds)
    rec = recall_score(y_test_tensor, preds, average="macro")
    f1 = f1_score(y_test_tensor, preds, average="macro")
    print(classification_report(y_test_tensor, preds, digits=4))

    # Manual garbage collection since RAM gets full
    del model
    gc.collect()
    return acc, rec, f1

vectorizers = {
    "CountVectorizer (BoW)": CountVectorizer(),
    "TF-IDF": TfidfVectorizer(),
    "Binary Vectorizer": CountVectorizer(binary=True),
    "BoW with N-grams": CountVectorizer(ngram_range=(1, 2)),
    "HashingVectorizer": HashingVectorizer(n_features=10000, alternate_sign=False)
}

results = []

for name, vectorizer in vectorizers.items():
    print(f"\nEvaluating {name}...")
    X_train_vec = vectorizer.fit_transform(X_train) if hasattr(vectorizer, "fit_transform") else vectorizer.transform(X_train)
    X_test_vec = vectorizer.transform(X_test)
    X_train_tensor = torch.tensor(X_train_vec.toarray(), dtype=torch.float32)
    X_test_tensor = torch.tensor(X_test_vec.toarray(), dtype=torch.float32)
    y_train_tensor = torch.tensor(y_train.values, dtype=torch.long)
    y_test_tensor = torch.tensor(y_test.values, dtype=torch.long)
    acc, rec, f1 = train_and_eval_nn(X_train_tensor, y_train_tensor, X_test_tensor, y_test_tensor, X_train_tensor.shape[1])
    results.append((name, acc, rec, f1))

    # Manual garbage collection since RAM gets full
    del X_train_vec, X_test_vec, X_train_tensor, X_test_tensor, y_train_tensor, y_test_tensor
    gc.collect()

print("\nAccuracy Scores for Different Vectorizers:")
print("========================================")
for name, acc, rec, f1 in results:
    print(f"{name}: Acc={acc:.4f}, Recall={rec:.4f}, F1={f1:.4f}")
```

```python
# ✅ Feed-Forward: Different Vectorizers (large NN)
# This cell compares different vectorizers for the larger of two feed-forward neural networks.
# increased epochs to 15 for better convergence
# roughly 1% better performance, but takes longer to train due to capacity and epochs
# suprisingly, n-grams do not outperform anymore

import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer, HashingVectorizer
from sklearn.metrics import accuracy_score, recall_score, f1_score, classification_report
import warnings
warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)

df_balanced = get_balanced_dataset()
df_balanced["label"] = df_balanced["sentiment"].astype("category").cat.codes

X_train, X_test, y_train, y_test = train_test_split(
    df_balanced["text"], df_balanced["label"],
    test_size=0.2, stratify=df_balanced["label"], random_state=SEED
)

class LargerFeedForwardNN(nn.Module):
    def __init__(self, input_dim, dropout=0.2):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, 256),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(128, 3)
        )
    def forward(self, x):
        return self.net(x)

def train_and_eval_nn(X_train_tensor, y_train_tensor, X_test_tensor, y_test_tensor, input_dim):
    model = LargerFeedForwardNN(input_dim)
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=0.001)
    model.train()
    for epoch in range(15):
        for i in range(0, len(X_train_tensor), 64):
            batch_x = X_train_tensor[i:i+64]
            batch_y = y_train_tensor[i:i+64]
            optimizer.zero_grad()
            outputs = model(batch_x)
            loss = criterion(outputs, batch_y)
            loss.backward()
            optimizer.step()
    model.eval()
    with torch.no_grad():
        preds = torch.argmax(model(X_test_tensor), dim=1)

    acc = accuracy_score(y_test_tensor, preds)
    rec = recall_score(y_test_tensor, preds, average="macro")
    f1 = f1_score(y_test_tensor, preds, average="macro")
    print(classification_report(y_test_tensor, preds, digits=4))

    # Manual garbage collection since RAM gets full
    del model
    gc.collect()
    return acc, rec, f1

vectorizers = {
    "CountVectorizer (BoW)": CountVectorizer(),
    "TF-IDF": TfidfVectorizer(),
    "Binary Vectorizer": CountVectorizer(binary=True),
    "BoW with N-grams": CountVectorizer(ngram_range=(1, 2)),
    "HashingVectorizer": HashingVectorizer(n_features=10000, alternate_sign=False)
}

results = []

for name, vectorizer in vectorizers.items():
    print(f"\nEvaluating {name}...")
    X_train_vec = vectorizer.fit_transform(X_train) if hasattr(vectorizer, "fit_transform") else vectorizer.transform(X_train)
    X_test_vec = vectorizer.transform(X_test)
    X_train_tensor = torch.tensor(X_train_vec.toarray(), dtype=torch.float32)
    X_test_tensor = torch.tensor(X_test_vec.toarray(), dtype=torch.float32)
    y_train_tensor = torch.tensor(y_train.values, dtype=torch.long)
    y_test_tensor = torch.tensor(y_test.values, dtype=torch.long)
    acc, rec, f1 = train_and_eval_nn(X_train_tensor, y_train_tensor, X_test_tensor, y_test_tensor, X_train_tensor.shape[1])
    results.append((name, acc, rec, f1))

    # Manual garbage collection since RAM gets full
    del X_train_vec, X_test_vec, X_train_tensor, X_test_tensor, y_train_tensor, y_test_tensor
    gc.collect()

print("\nAccuracy Scores for Different Vectorizers:")
print("========================================")
for name, acc, rec, f1 in results:
    print(f"{name}: Acc={acc:.4f}, Recall={rec:.4f}, F1={f1:.4f}")
```

```python
# ✅ Feed-Forward: Different Preprocessing (n-grams BoW)
# This cell compares different preprocessing options for both feed-forward neural networks.
# suprisingly, larger model benefits more from preprocessing
# larger models tend to perfom worse

import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import random
import numpy as np
import gc
import warnings

warnings.filterwarnings("ignore")

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

df = get_balanced_dataset()
df["label"] = df["sentiment"].astype("category").cat.codes

def prepare_data(df):
    X_train, X_test, y_train, y_test = train_test_split(
        df["text"], df["label"],
        test_size=0.2, stratify=df["label"], random_state=SEED
    )

    vectorizer = CountVectorizer(ngram_range=(1, 2))
    X_train_vec = vectorizer.fit_transform(X_train)
    X_test_vec = vectorizer.transform(X_test)
    X_train_tensor = torch.tensor(X_train_vec.toarray(), dtype=torch.float32)
    X_test_tensor = torch.tensor(X_test_vec.toarray(), dtype=torch.float32)
    y_train_tensor = torch.tensor(y_train.values, dtype=torch.long)
    y_test_tensor = torch.tensor(y_test.values, dtype=torch.long)

    # Manual garbage collection since RAM gets full
    del X_train, X_test, y_train, y_test
    gc.collect()
    return (
        X_train_tensor, X_test_tensor, y_train_tensor, y_test_tensor, 
        X_train_tensor.shape[1]
    )

class FeedForwardNN(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.fc = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.ReLU(),
            nn.Linear(128, 3)
        )
    def forward(self, x):
        return self.fc(x)
    
class LargerFeedForwardNN(nn.Module):
    def __init__(self, input_dim, dropout=0.2):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, 256),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(128, 3)
        )
    def forward(self, x):
        return self.net(x)

def train_and_eval(X_train, y_train, X_test, y_test, model):
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=0.001)
    model.train()
    for epoch in range(3):
        for i in range(0, len(X_train), 64):
            batch_x = X_train[i:i+64]
            batch_y = y_train[i:i+64]
            optimizer.zero_grad()
            outputs = model(batch_x)
            loss = criterion(outputs, batch_y)
            loss.backward()
            optimizer.step()
    model.eval()
    with torch.no_grad():
        preds = torch.argmax(model(X_test), dim=1)

    acc = accuracy_score(y_test, preds)

    # Manual garbage collection since RAM gets full
    del model, X_train, y_train, X_test, y_test, preds
    gc.collect()
    return acc

options = [
    "lowercase", "expand_contractions", "remove_urls", "emoticon_normalization",
    "detect_censored", "remove_mentions", "remove_punctuation", "preserve_ellipsis",
    "remove_numbers", "convert_numbers", "remove_non_ascii", "reduce_elongation",
    "remove_stopwords", "stemming", "lemmatization", "spelling_correction", "strip_multispace"
]

X_train_raw, X_test_raw, y_train_raw, y_test_raw, input_dim_raw = prepare_data(df)

model_base_small = FeedForwardNN(input_dim_raw)
acc_base_small = train_and_eval(X_train_raw, y_train_raw, X_test_raw, y_test_raw, model_base_small)

model_base_large = LargerFeedForwardNN(input_dim_raw)
acc_base_large = train_and_eval(X_train_raw, y_train_raw, X_test_raw, y_test_raw, model_base_large)

print(f"{'Base accuracy (small model)':30s}: {acc_base_small:.4f}")
print(f"{'Base accuracy (large model)':30s}: {acc_base_large:.4f}")
print()
print(f"{'Preprocessing Option':25s} | {'Small Δ':>8s} | {'Large Δ':>8s}")
print("-" * 50)

# Manual garbage collection since RAM gets full
del X_train_raw, X_test_raw, y_train_raw, y_test_raw, input_dim_raw, model_base_small, model_base_large
gc.collect()

for opt in options:
    kwargs = {k: False for k in options}
    kwargs[opt] = True
    df_pre = preprocessing(df.copy(), **kwargs)
    X_train_pre, X_test_pre, y_train_pre, y_test_pre, input_dim_pre = prepare_data(df_pre)

    model_small = FeedForwardNN(input_dim_pre)
    acc_small = train_and_eval(X_train_pre, y_train_pre, X_test_pre, y_test_pre, model_small)
    delta_small = acc_small - acc_base_small

    model_large = LargerFeedForwardNN(input_dim_pre)
    acc_large = train_and_eval(X_train_pre, y_train_pre, X_test_pre, y_test_pre, model_large)
    delta_large = acc_large - acc_base_large

    # Manual garbage collection since RAM gets full
    del df_pre, X_train_pre, X_test_pre, y_train_pre, y_test_pre, input_dim_pre, model_small, model_large
    gc.collect()

    print(f"{opt:25s} | {delta_small:+.4f} | {delta_large:+.4f}")
```

```python
# ✅ Feed-Forward: Error Analysis

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
import gc
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import accuracy_score, recall_score, f1_score, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)

df_balanced = get_balanced_dataset()
df_balanced["label"] = df_balanced["sentiment"].astype("category").cat.codes
label_to_sentiment = dict(enumerate(df_balanced["sentiment"].astype("category").cat.categories))

X_train, X_test, y_train, y_test = train_test_split(
    df_balanced["text"], df_balanced["label"],
    test_size=0.2, stratify=df_balanced["label"], random_state=SEED
)

class FeedForwardNN(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.fc = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.ReLU(),
            nn.Linear(128, 3)
        )
    def forward(self, x):
        return self.fc(x)

vectorizer = CountVectorizer(ngram_range=(1, 2))
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

X_train_tensor = torch.tensor(X_train_vec.toarray(), dtype=torch.float32)
X_test_tensor = torch.tensor(X_test_vec.toarray(), dtype=torch.float32)
y_train_tensor = torch.tensor(y_train.values, dtype=torch.long)
y_test_tensor = torch.tensor(y_test.values, dtype=torch.long)

model = FeedForwardNN(X_train_tensor.shape[1])
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
model.train()

for epoch in range(10):
    for i in range(0, len(X_train_tensor), 64):
        batch_x = X_train_tensor[i:i+64]
        batch_y = y_train_tensor[i:i+64]
        optimizer.zero_grad()
        outputs = model(batch_x)
        loss = criterion(outputs, batch_y)
        loss.backward()
        optimizer.step()

model.eval()
with torch.no_grad():
    logits = model(X_test_tensor)
    probs = torch.softmax(logits, dim=1)
    preds = torch.argmax(probs, dim=1)

acc = accuracy_score(y_test_tensor, preds)
rec = recall_score(y_test_tensor, preds, average="macro")
f1 = f1_score(y_test_tensor, preds, average="macro")
print(f"\nAccuracy: {acc:.4f}, Recall: {rec:.4f}, F1: {f1:.4f}")

confidences = torch.max(probs, dim=1).values.numpy()
errors_df = pd.DataFrame({
    "text": X_test.values,
    "true_label": y_test.values,
    "predicted_label": preds.numpy(),
    "confidence": confidences
})
errors_df["correct"] = errors_df["true_label"] == errors_df["predicted_label"]
errors_only_df = errors_df[~errors_df["correct"]].copy()
errors_only_df["true_label"] = errors_only_df["true_label"].map(label_to_sentiment)
errors_only_df["predicted_label"] = errors_only_df["predicted_label"].map(label_to_sentiment)
errors_only_df = errors_only_df.sort_values(by="confidence", ascending=False)

cm = confusion_matrix(y_test_tensor, preds)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=[label_to_sentiment[i] for i in range(3)])
disp.plot(cmap="Blues")
plt.title("Confusion Matrix: Feed-Forward NN with N-grams BoW")
plt.show()
```

```python
# ✅ Binary Classification vs. Multiclass Classification

import gc
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)

df = get_balanced_dataset()

def run_pipeline(df, title_prefix):
    df["label"] = df["sentiment"].astype("category").cat.codes
    X_train, X_test, y_train, y_test = train_test_split(
        df["text"], df["label"], test_size=0.2, stratify=df["label"], random_state=SEED
    )
    vectorizer = CountVectorizer(ngram_range=(1, 2))
    X_train_vec = vectorizer.fit_transform(X_train)
    X_test_vec = vectorizer.transform(X_test)

    nb_model = MultinomialNB()
    nb_model.fit(X_train_vec, y_train)
    y_pred_nb = nb_model.predict(X_test_vec)

    print(f"\n--- {title_prefix} Naive Bayes ---")
    print(classification_report(y_test, y_pred_nb, digits=4))
    ConfusionMatrixDisplay.from_predictions(y_test, y_pred_nb)
    plt.title(f"{title_prefix} Naive Bayes Confusion Matrix")
    plt.show()

    del vectorizer, nb_model, y_pred_nb
    gc.collect()

    class FeedForwardNN(nn.Module):
        def __init__(self, input_dim, output_dim):
            super().__init__()
            self.fc = nn.Sequential(
                nn.Linear(input_dim, 128),
                nn.ReLU(),
                nn.Linear(128, output_dim)
            )
        def forward(self, x):
            return self.fc(x)

    X_train_tensor = torch.tensor(X_train_vec.toarray(), dtype=torch.float32)
    X_test_tensor = torch.tensor(X_test_vec.toarray(), dtype=torch.float32)
    y_train_tensor = torch.tensor(y_train.values, dtype=torch.long)
    y_test_tensor = torch.tensor(y_test.values, dtype=torch.long)

    nn_model = FeedForwardNN(X_train_tensor.shape[1], df["label"].nunique())
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(nn_model.parameters(), lr=0.001)

    nn_model.train()
    for epoch in range(10):
        for i in range(0, len(X_train_tensor), 64):
            batch_x = X_train_tensor[i:i+64]
            batch_y = y_train_tensor[i:i+64]
            optimizer.zero_grad()
            outputs = nn_model(batch_x)
            loss = criterion(outputs, batch_y)
            loss.backward()
            optimizer.step()

    nn_model.eval()
    with torch.no_grad():
        outputs = nn_model(X_test_tensor)
        y_pred_nn = torch.argmax(outputs, dim=1)

    print(f"\n--- {title_prefix} Feed-Forward NN ---")
    print(classification_report(y_test_tensor, y_pred_nn, digits=4))
    ConfusionMatrixDisplay.from_predictions(y_test_tensor, y_pred_nn)
    plt.title(f"{title_prefix} Feed-Forward NN Confusion Matrix")
    plt.show()

    del nn_model, criterion, optimizer

df_binary = df[df["sentiment"].isin(["positive", "negative"])].copy()
run_pipeline(df_binary, "Binary")

df_multiclass = df.copy()
run_pipeline(df_multiclass, "Multiclass")
```
